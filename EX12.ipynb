```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("test3.jpg", 0)

A = 50
B = 150

slice_without_bg = np.zeros_like(image)
slice_without_bg[(image >= A) & (image <= B)] = 255

slice_with_bg = image.copy()
slice_with_bg[(image >= A) & (image <= B)] = 255

def intensity_slice(img, low, high):
    result = np.zeros_like(img)
    result[(img >= low) & (img <= high)] = 255
    return result

slice_02 = intensity_slice(image, 0, int(0.2 * 255))
slice_04 = intensity_slice(image, int(0.2 * 255), int(0.4 * 255))
slice_06 = intensity_slice(image, int(0.4 * 255), int(0.6 * 255))
slice_08 = intensity_slice(image, int(0.6 * 255), int(0.8 * 255))

plt.figure(figsize=(10,8))

plt.subplot(3,2,1)
plt.title("Original Image")
plt.imshow(image, cmap='gray')
plt.axis('off')

plt.subplot(3,2,2)
plt.title("Binary Slice")
plt.imshow(slice_without_bg, cmap='gray')
plt.axis('off')

plt.subplot(3,2,3)
plt.title("0.2 Slice")
plt.imshow(slice_02, cmap='gray')
plt.axis('off')

plt.subplot(3,2,4)
plt.title("0.4 Slice")
plt.imshow(slice_04, cmap='gray')
plt.axis('off')

plt.subplot(3,2,5)
plt.title("0.6 Slice")
plt.imshow(slice_06, cmap='gray')
plt.axis('off')

plt.subplot(3,2,6)
plt.title("0.8 Slice")
plt.imshow(slice_08, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```
